# GINE vs GAT and Mean vs Max Pooling Comparison
This notebook compares the full test evaluation performance and training curves of `gs_s3p` runs against the `gs_s2_e0_n0_v0` baseline (GINE with mean pooling). All runs evaluated on seed 0.

```python

import json
import pandas as pd
import plotly.express as px
from pathlib import Path

eval_dir = Path("/Users/corentinplumet/Documents/RL_Marl2grid/Topology_Task/outputs/full_test_eval")

files = {
    "GINE (Mean Pooling)": eval_dir / "best_test_gs_s2_bus_n0_none_e0n0v0_s0_step14100480_job65858375.json",
    "GINE (Max Pooling)": eval_dir / "gs_s3p" / "gs_s3ep_bus_n0_none_e0n0v0_gine_max_s0.json",
    "GAT (Mean Pooling)": eval_dir / "gs_s3p" / "gs_s3ep_bus_n0_none_e0n0v0_gat_mean_s0.json",
    "GAT (Max Pooling)": eval_dir / "gs_s3p" / "gs_s3ep_bus_n0_none_e0n0v0_gat_max_s0.json",
}

results = []
for label, path in files.items():
    if path.exists():
        with open(path) as f:
            data = json.load(f)
            arch = label.split(" ")[0]
            pool = label.split("(")[1].split(" ")[0]
            results.append({
                "Configuration": label,
                "Architecture": arch,
                "Pooling": pool,
                "Episodic Survival (%)": data.get("survival_percent", 0.0)
            })
    else:
        print(f"File not found: {path}")

df = pd.DataFrame(results)
display(df)
```

```python

fig1 = px.bar(
    df, 
    x="Configuration", 
    y="Episodic Survival (%)",
    color="Architecture",
    pattern_shape="Pooling",
    title="Full Test Episodic Survival: GINE vs GAT and Mean vs Max Pooling (Seed 0)",
    text_auto=".2f",
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig1.update_layout(
    yaxis=dict(range=[0, 105]),
    width=800,
    height=500
)
fig1.show()
```
